```python
# ============================================
# STEP 4: CREATE TRAIN AND TEST DATASETS
# ============================================

# Get the original dataset
full_dataset = dataset["train"]

# Separate using the existing split column
train_data = full_dataset.filter(
    lambda x: x["split"] == "train"
)

test_data = full_dataset.filter(
    lambda x: x["split"] == "test"
)

print("Training images:", len(train_data))
print("Testing images :", len(test_data))

print("\nTraining classes:", len(set(train_data["class_label"])))
print("Testing classes :", len(set(test_data["class_label"])))
```

```python
# ============================================
# STEP 5: CLASS MAPPING
# ============================================

classes = sorted(set(train_data["class_label"]))

label2id = {
    label: idx
    for idx, label in enumerate(classes)
}

id2label = {
    idx: label
    for idx, label in enumerate(classes)
}

print("Number of classes:", len(classes))

for idx, label in id2label.items():
    print(idx, ":", label)
```

```python
# ============================================
# STEP 6: LOAD PRETRAINED VISION TRANSFORMER
# ============================================

from transformers import AutoImageProcessor
from transformers import AutoModelForImageClassification

model_name = "google/vit-base-patch16-224"

processor = AutoImageProcessor.from_pretrained(
    model_name
)

model = AutoModelForImageClassification.from_pretrained(
    model_name,
    num_labels=len(classes),
    id2label=id2label,
    label2id=label2id,
    ignore_mismatched_sizes=True
)

print("Model loaded successfully!")
print("Number of output classes:", model.config.num_labels)
```

```python
print("Image size:", processor.size)
print("Image mean:", processor.image_mean)
print("Image std :", processor.image_std)
```

```python
from torchvision.transforms import (
    Compose,
    RandomResizedCrop,
    RandomHorizontalFlip,
    RandomRotation,
    ColorJitter,
    Resize,
    CenterCrop,
    ToTensor,
    Normalize
)

image_mean = processor.image_mean
image_std = processor.image_std

# Training augmentation
train_transform = Compose([
    RandomResizedCrop(224, scale=(0.8, 1.0)),
    RandomHorizontalFlip(p=0.5),
    RandomRotation(10),
    ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),
    ToTensor(),
    Normalize(
        mean=image_mean,
        std=image_std
    )
])

# Test preprocessing
test_transform = Compose([
    Resize(256),
    CenterCrop(224),
    ToTensor(),
    Normalize(
        mean=image_mean,
        std=image_std
    )
])

print("Preprocessing created successfully!")
```

```python
def train_preprocess(examples):

    images = [
        image.convert("RGB")
        for image in examples["image"]
    ]

    pixel_values = [
        train_transform(image)
        for image in images
    ]

    return {
        "pixel_values": pixel_values,
        "labels": examples["class_idx"]
    }


def test_preprocess(examples):

    images = [
        image.convert("RGB")
        for image in examples["image"]
    ]

    pixel_values = [
        test_transform(image)
        for image in images
    ]

    return {
        "pixel_values": pixel_values,
        "labels": examples["class_idx"]
    }

print("Preprocessing functions created!")
```

```python
train_dataset = train_data.with_transform(
    train_preprocess
)

test_dataset = test_data.with_transform(
    test_preprocess
)

print("Training dataset:", len(train_dataset))
print("Testing dataset :", len(test_dataset))
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./crop_disease_vit",

    # Evaluation and saving
    eval_strategy="epoch",
    save_strategy="epoch",

    # Learning parameters
    learning_rate=2e-5,
    weight_decay=0.01,

    # Batch sizes
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,

    # Training duration
    num_train_epochs=3,

    # Logging
    logging_steps=50,

    # Keep best model
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",
    greater_is_better=True,

    # GPU optimization
    fp16=True,

    # Avoid external logging
    report_to="none",

    # Keep only best few checkpoints
    save_total_limit=2
)

print("Training configuration created successfully!")
```

```python
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

def compute_metrics(eval_pred):

    predictions, true_labels = eval_pred

    predicted_labels = np.argmax(
        predictions,
        axis=1
    )

    accuracy = accuracy_score(
        true_labels,
        predicted_labels
    )

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            true_labels,
            predicted_labels,
            average="weighted",
            zero_division=0
        )
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("Evaluation metrics ready!")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,

    train_dataset=train_dataset,
    eval_dataset=test_dataset,

    processing_class=processor,

    compute_metrics=compute_metrics
)

print("Trainer created successfully!")
```

```python
# ============================================
# RECREATE TEST DATASET
# ============================================

from datasets import load_dataset
from torchvision.transforms import (
    Compose,
    Resize,
    CenterCrop,
    ToTensor,
    Normalize
)

# 1. Load PlantVillage
dataset = load_dataset(
    "geraldmc/plantvillage-full",
    revision="v0.1.0"
)

# 2. Get the full dataset
full_dataset = dataset["train"]

# 3. Select the official test split
test_data = full_dataset.filter(
    lambda x: x["split"] == "test"
)

print("Test images:", len(test_data))

# 4. Test preprocessing
test_transform = Compose([
    Resize(256),
    CenterCrop(224),
    ToTensor(),
    Normalize(
        mean=processor.image_mean,
        std=processor.image_std
    )
])

# 5. Create preprocessing function
def test_preprocess(examples):

    images = [
        image.convert("RGB")
        for image in examples["image"]
    ]

    pixel_values = [
        test_transform(image)
        for image in images
    ]

    return {
        "pixel_values": pixel_values,
        "labels": examples["class_idx"]
    }

# 6. Create model-ready test dataset
test_dataset = test_data.with_transform(
    test_preprocess
)

print("Test dataset ready!")
print("Number of test images:", len(test_dataset))
```

```python
sample = test_dataset[0]

print("Keys:", sample.keys())
print("Image shape:", sample["pixel_values"].shape)
print("Actual class ID:", sample["labels"])
```

```python
import torch
import numpy as np
from torch.utils.data import DataLoader
from tqdm.auto import tqdm

# Use GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model.to(device)
model.eval()

# Batch collator
def collate_fn(batch):
    pixel_values = torch.stack([
        item["pixel_values"] for item in batch
    ])

    labels = torch.tensor([
        item["labels"] for item in batch
    ])

    return {
        "pixel_values": pixel_values,
        "labels": labels
    }

# DataLoader
test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    collate_fn=collate_fn,
    num_workers=2,
    pin_memory=True
)

all_predictions = []
all_labels = []

print("Starting evaluation...")
print("Test images:", len(test_dataset))

with torch.no_grad():

    for batch in tqdm(test_loader):

        pixel_values = batch["pixel_values"].to(
            device,
            non_blocking=True
        )

        labels_batch = batch["labels"]

        outputs = model(
            pixel_values=pixel_values
        )

        predictions = torch.argmax(
            outputs.logits,
            dim=1
        ).cpu().numpy()

        all_predictions.extend(predictions)
        all_labels.extend(labels_batch.numpy())

all_predictions = np.array(all_predictions)
all_labels = np.array(all_labels)

print("\n======================================")
print("EVALUATION COMPLETED")
print("======================================")
print("Total predictions:", len(all_predictions))
print("Total labels:", len(all_labels))
```

```python
from google.colab import files
from PIL import Image
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F

# ============================================
# UPLOAD IMAGE
# ============================================

print("Upload a leaf image to test the model:")

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

image = Image.open(image_path).convert("RGB")


# ============================================
# PREPROCESS IMAGE
# ============================================

pixel_values = test_transform(image).unsqueeze(0).to(device)


# ============================================
# MODEL PREDICTION
# ============================================

model.eval()

with torch.no_grad():

    outputs = model(
        pixel_values=pixel_values
    )

    probabilities = F.softmax(
        outputs.logits,
        dim=-1
    )[0]


# ============================================
# GET TOP PREDICTION
# ============================================

predicted_id = torch.argmax(probabilities).item()

predicted_label = model.config.id2label[predicted_id]

predicted_confidence = probabilities[predicted_id].item()


# ============================================
# SPLIT CROP AND CONDITION
# ============================================

if "___" in predicted_label:

    species_raw, condition_raw = predicted_label.split(
        "___",
        1
    )

else:

    species_raw = predicted_label
    condition_raw = "Unknown"


# Clean names

species = species_raw.replace("_", " ").strip()

condition = condition_raw.replace("_", " ").strip()


# ============================================
# DISPLAY IMAGE
# ============================================

plt.figure(figsize=(7, 7))

plt.imshow(image)

plt.axis("off")

plt.title(
    f"Species: {species}\n"
    f"Condition: {condition}",
    fontsize=14
)

plt.show()


# ============================================
# FINAL OUTPUT
# ============================================

print()
print("=" * 55)
print("🌱  CROP DISEASE DETECTION")
print("=" * 55)

print(
    f"Species: {species} "
    f"({predicted_confidence * 100:.2f}%)"
)

print(
    f"Condition: {condition} "
    f"({predicted_confidence * 100:.2f}%)"
)

print("=" * 55)

print({
    "species": species,
    "species_confidence": round(
        predicted_confidence,
        4
    ),
    "condition": condition,
    "condition_confidence": round(
        predicted_confidence,
        4
    ),
    "predicted_class": predicted_label
})
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F
import numpy as np

# Load current image
image = Image.open(image_path).convert("RGB")

print("Image size:", image.size)

# Create several crops
w, h = image.size

crops = {
    "Full Image": image,
    "Top": image.crop((0, 0, w, int(h * 0.55))),
    "Middle": image.crop((0, int(h * 0.20), w, int(h * 0.80))),
    "Bottom": image.crop((0, int(h * 0.45), w, h)),
}

results = []

model.eval()

for crop_name, crop in crops.items():

    pixel_values = test_transform(crop).unsqueeze(0).to(device)

    with torch.no_grad():
        outputs = model(pixel_values=pixel_values)

    probs = F.softmax(outputs.logits, dim=-1)[0]

    top_prob, top_id = torch.max(probs, dim=0)

    results.append({
        "crop": crop_name,
        "class": model.config.id2label[top_id.item()],
        "confidence": top_prob.item()
    })


print("\n========================================")
print("MULTI-CROP TEST")
print("========================================")

for result in results:
    print(
        f"{result['crop']:12} → "
        f"{result['class']:55} "
        f"{result['confidence'] * 100:.2f}%"
    )

print("========================================")
```

## SECOND VERSION OF THIS MODEL FOR REAL WORLD IMAGES

```python
from PIL import Image
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F

# ============================================
# LOAD THE CURRENT FIELD IMAGE
# ============================================

image = Image.open(image_path).convert("RGB")

w, h = image.size

print("Original image size:", image.size)


# ============================================
# CREATE LEAF-FOCUSED CROPS
# ============================================

# Center square crop
crop_size = min(w, h)

left = (w - crop_size) // 2
top = (h - crop_size) // 2

center_crop = image.crop(
    (left, top, left + crop_size, top + crop_size)
)

# Middle region
middle_crop = image.crop(
    (
        0,
        int(h * 0.20),
        w,
        int(h * 0.80)
    )
)

# Upper-middle region
upper_crop = image.crop(
    (
        0,
        int(h * 0.10),
        w,
        int(h * 0.65)
    )
)


crops = {
    "Full Image": image,
    "Center Leaf": center_crop,
    "Middle Leaves": middle_crop,
    "Upper Leaves": upper_crop
}


# ============================================
# DISPLAY CROPS
# ============================================

fig, axes = plt.subplots(
    1,
    4,
    figsize=(16, 5)
)

for ax, (name, crop) in zip(axes, crops.items()):

    ax.imshow(crop)
    ax.set_title(name)
    ax.axis("off")

plt.tight_layout()
plt.show()


# ============================================
# PREDICT EACH CROP
# ============================================

model.eval()

print("\n==============================================")
print("LEAF-FOCUSED TEST")
print("==============================================")

for crop_name, crop in crops.items():

    pixel_values = test_transform(
        crop
    ).unsqueeze(0).to(device)

    with torch.no_grad():

        outputs = model(
            pixel_values=pixel_values
        )

        probabilities = F.softmax(
            outputs.logits,
            dim=-1
        )[0]

    top_prob, top_id = torch.max(
        probabilities,
        dim=0
    )

    predicted_class = model.config.id2label[
        top_id.item()
    ]

    print(
        f"{crop_name:16} → "
        f"{predicted_class:55} "
        f"{top_prob.item()*100:.2f}%"
    )

print("==============================================")
```
